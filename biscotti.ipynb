```python
import os
import re
import requests

from bs4 import BeautifulSoup
from datetime import datetime
```

```python
#Base Page Url
BASE_PAGE_URL = "https://www.giallozafferano.it/ricette-cat/Dolci-e-Desserts/Biscotti/"
#"https://www.giallozafferano.it/ricette-cat/page{}/Dolci-e-Desserts/Biscotti/"

#List to contain cookie data
cookieData = []

#Get current date
timestamp = datetime.now()

#Output folder and files
output_dir= "rawHtml_" + timestamp.strftime("%Y-%m-%d")

#Debug
#print(timestamp)
#print(date)
#print(output_dir)
```

```python
#DONE : -function03- Write here a function to iterate through pages and call function01 on each.
#DONE / TOTEST : 2025-09-22 | #TESTED : 2025-09-29 | #DONE / DONE : 2025-09-29
def scrapeCookiesPagesTotal(BASE_PAGE_URL):
  #Request base Page
  totalPageNumber = None
  try: 
    response = requests.get(BASE_PAGE_URL, timeout=10)
    response.raise_for_status()
    print(f"Request Done Successfully (Status Code : {response.status_code})...")
    #print(f"Request Done Successfully (Status Code :" + {response.status_code} + ")...")
    
    #Get Total Number of Pages
    soupPageNumber = BeautifulSoup(response.text, "html.parser")
    span_tag = soupPageNumber.find("span", {"class": "disabled total-pages"})

    if span_tag:
      totalPageNumber = span_tag.get_text(strip=True)
      print("Base Page Souped Successfully...")
    else:
      print("Total Number of Pages Not Found...")

   #Raise Error If Any. 
  except requests.exceptions.HTTPError as e:
        print("Request Failed:", e)
  except requests.exceptions.RequestException as e:
        print(f"Error Fetching {BASE_PAGE_URL}: {e}")

  return totalPageNumber

#BASE_PAGE_URL = "https://www.giallozafferano.it/ricette-cat/Dolci-e-Desserts/Biscotti/"
#print(scrapeCookiesPagesTotal(BASE_PAGE_URL))
```

```python
#DONE / TOTEST : 2025-09-22 | #TESTED : 2025-09-22 | #DONE / DONE : 2025-09-22
### THIS FUNCTION BUILD DIFFERENTE PAGE URL SO THEY CAN BE ITERATED OVER
#PAGE_URL, page = "https://www.giallozafferano.it/ricette-cat/Dolci-e-Desserts/Biscotti/", 3 - DEBUG
def pageUrlBuilder(BASE_PAGE_URL, page) :
  print(f"Building Rrl For Page nº{page}...")
  parts = re.split(r"(ricette-cat/)", BASE_PAGE_URL, maxsplit=1)  
  return f"{parts[0]}{parts[1]}page{page}/{parts[2]}"
#print(pageUrlBuilder(PAGE_URL, page)) - DEBUG
```

```python
#DONE / TOTEST : 2025-10-01 | #TESTED : 2025-10-01 | #DONE / DONE : 2025-10-01
#TODO -function00- Use Requests to extract hml / and store soup page per page
#function gets 
def requestCookiePages(BASE_PAGE_URL, totalPageNumber, output_dir, timestamp):
  page = 1 #for debug
  os.makedirs(output_dir, exist_ok=True)

  safe_timestamp = str(timestamp).replace(":", "-").replace(" ", "_").replace(".", "-")

  for page in range(1, totalPageNumber + 1):
    pageUrls = pageUrlBuilder(BASE_PAGE_URL, page)
    #print(pageUrls)
    try: 
      response = requests.get(pageUrls, timeout=10)
      response.raise_for_status()

      # Save HTML to file
      filename = os.path.join(output_dir, f"page_{page}_of_{totalPageNumber}_{safe_timestamp}.html")
      with open(filename, "w", encoding="utf-8") as f:
          f.write(response.text)
      
      print(f"Saved page {page} to {filename}...")

    except requests.exceptions.HTTPError as e:
      print("Request failed:", e)
    except requests.exceptions.RequestException as e:
      print(f"Error fetching {pageUrls}: {e}")

#totalPageNumber = 14 
#requestCookiePages(BASE_PAGE_URL, totalPageNumber, output_dir, timestamp)
```
