```python
#TODO : -function03- Write here a function to iterate throught pages and call function01 on each.
#DONE / TOTEST : 2025-09-22 
#TESTED : ?
#DONE / DONE : ?
def scrapeCookiesPagesTotal(PAGE_URL):
      #Request base Page
      try: 
        response = requests.get(PAGE_URL, timeout=10)
        response.raise_for_status()
        print("Request Done Successfully (Status Code :" + response.status_code + ")...")

      #Raise Error If Any. 
      except requests.exceptions.HTTPError as e:
        print("Request Failed:", e)
      except requests.exceptions.RequestException as e:
        print(f"Error Fetching {PAGE_URL}: {e}")
        
        #Get Total Number of Pages
        soupPageNumber = BeautifulSoup(response.text, "html.parser")
        totalPageNumber = soupPageNumber.find_all("span", "class" : "disabled total-pages").get_text(strip=True)
        if totalPageNumber : 
          print("Base Page Souped Successfully...")
        else:
           print("Total Number of Pages Not Found...")

        #Build Each Page Url. Soup data from them All. Append to sink list.
        for page in (totalPageNumber):
          pageUrls = pageUrlBuilder(PAGE_URL, page)
          cookieData = requestCookiePages(pageUrls)
          cookieData.append(cookieData)
          print(f"Page nº{page}/{totalPageNumber} Souped Successfully...")

        print("Souping Done...")

        print("Building List...")
        return cookieData
```
